```python
##Check numerical columns
numeric_columns = [
    'Discount',
    'Profit',
    'Quantity',
    'Sales',
    'Shipping Cost',
    'Year',
    'weeknum'
]

df[numeric_columns].dtypes
```

```python
for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
##Check categorical columns
categorical_columns = [
    'Category',
    'City',
    'Country',
    'Market',
    'Order Priority',
    'Region',
    'Segment',
    'Ship Mode',
    'State',
    'Sub-Category',
    'Market2'
]

for col in categorical_columns:
    df[col] = df[col].astype('category')
```
